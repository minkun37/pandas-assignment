## 5. 복합 조건

각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
mask = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] == 1)
print(sum(mask))
```

```python
mask = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
print(sum(mask))
```

```python
mask = ~(titanic["alone"] == True)
print(sum(mask))
```

## 6. isin · between

1. `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
2. `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.

```python
mask = titanic["embark_town"].isin(["Cherbourg" , "Queenstown"])
emb_mask = titanic[mask]
print(emb_mask.shape)
```

```python
mask = titanic["age"].between(20 , 30)
age_mask = titanic[mask]
print(age_mask.shape)
print(age_mask["age"].min())
print(age_mask["age"].max())
```

```python
mask = titanic["age"].between(20 , 30 , inclusive="neither")
age_mask_n = titanic[mask]
print(age_mask_n.shape)
```

## 7. loc[조건, 열]

12살 이하 **그리고** 3등석 승객의 `age`, `sex`, `survived` 열만 `kids` 변수에 담고, 행 수와 생존률(`survived` 평균, 소수 둘째 자리)을 출력하세요.

```python
mask = (titanic["age"] <= 12) & (titanic["pclass"] == 3)
kids = titanic.loc[mask , ["age" , "sex" , "survived"]]
print(sum(mask))

kids_mean = kids["survived"].mean() * 100
print(kids_mean.round(2))
```

## 8. query

1. 변수 `min_fare = 50`을 만들고, `query`로 Cherbourg 탑승 **그리고** 요금이 `min_fare` 이상인 승객의 `shape`를 출력하세요.
2. 5-1번 조건을 `query`로 다시 써서 행 수가 같은지 확인하세요.

## 9. 정렬

1. `pclass` 오름차순, 같은 등급 안에서는 `fare` 내림차순으로 정렬해 `pclass`, `fare`, `sex`, `age` 열의 상위 3행을 출력하세요.
2. 요금 내림차순으로 정렬한 결과를 변수에 담고, `iloc[0]`과 `loc[0]`의 `fare` 값을 각각 출력하세요. 두 값이 다른 이유를 주석으로 쓰세요.

```python
pf_sort = titanic.sort_values(
    by=["pclass" , "fare"] ,
    ascending=[True , False]
)
pf_sort_pfsa = pf_sort[ ["pclass" , "fare" , "sex" , "age"] ].head(3)
print(pf_sort_pfsa)
```

```python
fare_sort = titanic.sort_values(
    by=["fare"] ,
    ascending=[False]
)
print(fare_sort.iloc[0]["fare"])
print(fare_sort.loc[0]["fare"])

# iloc[0]은 정수 위치를 가져오므로 해당 DataFrame의 0번째 행을 가져오고, loc[0]은 index의 이름을 가져오므로 이름이 0인 행을 가져옴
```

## 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
mask = (titanic["sex"] == "male") & (titanic["age"].between(20 , 40)) & (titanic["pclass"].isin([1 , 2]))
sap_mask = titanic[mask]
sap_mask[["pclass" , "age" , "fare" , "survived"]]
```

```python
print(sum(mask))

sap_mean = sap_mask["survived"].mean() * 100
print(sap_mean.round(2))

sap_mask_sort = sap_mask.sort_values(by="fare" , ascending=False).head(3)
print(sap_mask_sort)
```

```python
sap_qr = titanic.query("sex == 'male' and age >= 20 and age <= 40 and (pclass == 1 | pclass == 2)")
sap_qr.shape[0]

# 103행으로 동일함
```

```python
sap_qr_b = titanic.query("sex == 'male' and age >= 20 and age <= 40 and pclass == 3")
sap_mean_b = sap_qr_b["survived"].mean() * 100
print(sap_mean_b.round(2))
```
