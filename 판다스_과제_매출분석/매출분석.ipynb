```python
import glob

# data/baemin 폴더에서 baemin으로 시작하는 csv 파일 경로를 모두 찾아 리스트로 돌려줌
# * 는 "아무 글자나"라는 뜻 → baemin20261001.csv, baemin20261002.csv ... 전부 찾음
# sorted() 로 날짜 순서대로 정렬
baemin_files = sorted(glob.glob("data/baemin/baemin*.csv"))
baemin_files
```

```python
import pandas as pd
import seaborn as sns

baemin_list = []
for file in baemin_files:
    df = pd.read_csv(file , encoding="utf-8")
    baemin_list.append(df)

baemin_df = pd.concat(baemin_list , ignore_index=True)
baemin_df
```

```python
coupang_files = sorted(glob.glob("data/coupang/coupang*.csv"))
coupang_files
```

```python
coupang_list = []
for file in coupang_files:
    df = pd.read_csv(file , encoding="utf-8")
    coupang_list.append(df)

coupang_df = pd.concat(coupang_list , ignore_index=True)
coupang_df
```

```python
baemin_df.columns = ["주문일시" , "전체금액" , "광고유입여부" , "cpc"]
baemin_df
```

```python
coupang_df.columns = ["주문일시" , "전체금액" , "광고유입여부" , "ad_pct"]
coupang_df
```

```python
baemin_df["플랫폼"] = "배민"
baemin_df
```

```python
coupang_df["플랫폼"] = "쿠팡"
coupang_df
```

```python
baemin_df["광고집행비"] = baemin_df["cpc"]
baemin_df
```

```python
coupang_df["광고집행비"] = coupang_df["전체금액"] * coupang_df["ad_pct"] / 100
coupang_df
```

```python
df = pd.concat([baemin_df , coupang_df] , ignore_index=True)
df
```

```python
df["주문일시"] = pd.to_datetime(df["주문일시"])
df["주문일자"] = df["주문일시"].dt.day
df
```

```python
df = df.sort_values(by="주문일시" , ascending=True)
df
```

```python
amount_table = df.pivot_table(
    index="주문일자" ,
    columns="플랫폼" ,
    values="전체금액" ,
    aggfunc="sum" ,
)
amount_table
```

```python
amount_table["합계"] = amount_table["배민"] + amount_table["쿠팡"]
amount_table
```

```python
ad_mask = df["광고유입여부"] == "Y"
ad_mask_pct = ad_mask.mean() * 100
print(ad_mask_pct.round(2))
```

```python
bm_mask = df["플랫폼"] == "배민"
bm_ad = df[bm_mask]["광고집행비"].sum()

cp_mask = df["플랫폼"] == "쿠팡"
cp_ad = df[cp_mask]["광고집행비"].sum()

print(bm_ad)
print(cp_ad)
```

```python
bank = pd.read_csv("data/bank_20261001_20261031.csv" ,  encoding="cp949")
bank
```

```python
ing_mask = bank["거래처"] == "식자재마트"
ing_wit = bank[ing_mask]["출금액"]
ing_wit
```

```python
bank["거래일시"] = pd.to_datetime(bank["거래일시"])
bank["거래일자"] = bank["거래일시"].dt.day
bank
```

```python
ing_table = bank.pivot_table(
    index="거래일자" ,
    columns="거래처" ,
    values="출금액" ,
    aggfunc="sum" ,
)
ing_table["식자재마트"] = ing_table["식자재마트"].astype(int)
ing_table["식자재마트"]
```

```python
amount_table["식자재비"] = ing_table["식자재마트"]
amount_table["매출-식자재비"] = amount_table["합계"] - amount_table["식자재비"]
amount_table = amount_table.reset_index()
amount_table
```

```python
amount_table.to_csv("data/result.csv" , index=False , encoding="utf-8-sig")
```
